```python
from ast import Add
from pydantic import BaseModel


class Address(BaseModel):
  city : str
  state : str
  pin : int

class Patient(BaseModel):

  name : str
  gender : str
  age : int
  address : Address

address_dict = {'city':'Nagpur', 'state':'Maharashtra', 'pin': 441111}

address1 = Address(**address_dict)

patient_dict = {'name':'Ansul', 'gender':'male', 'age':23, 'address': address1}

patient1 = Patient(**patient_dict)

temp = patient1.model_dump()
print(temp)
print(type(temp))
```

```python
# exclude unset
from ast import Add
from pydantic import BaseModel


class Address(BaseModel):
  city : str
  state : str
  pin : int

class Patient(BaseModel):

  name : str
  gender : str = 'male'
  age : int
  address : Address

address_dict = {'city':'Nagpur', 'state':'Maharashtra', 'pin': 441111}

address1 = Address(**address_dict)

patient_dict = {'name':'Ansul', 'age':23, 'address': address1}

patient1 = Patient(**patient_dict)

temp = patient1.model_dump(exclude_unset=True) # object banate time jo chize set nhi hui hai wo export nhi hogi example gender

print(temp)
print(type(temp))
```
