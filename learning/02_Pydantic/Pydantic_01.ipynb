* Custom data validation : Field Function

```python
from pydantic import BaseModel, EmailStr, AnyUrl, Field
from typing import List, Dict

class Patient(BaseModel):

  name : str = Field(max_length=50) # 50 max character
  email : EmailStr
  linkedin_url : AnyUrl
  age : int = Field(gt=18, lt=48) # greater than 18 and less then 48
  weight : float = Field(gt=0)
  married : bool
  allergies : List[str] = Field(max_length=5) # no more then 5 allergies can be added
  contact_details : Dict[str, str]


def insert_patient_data(patient: Patient):

  print(patient.name)
  print(patient.email)
  print(patient.linkedin_url)
  print(patient.age)
  print(patient.weight)
  print(patient.married)
  print(patient.allergies)
  print(patient.contact_details)
  print('Inserted into database')

def update_patient_data(patient: Patient):

  print(patient.name)
  print(patient.email)
  print(patient.age)
  print(patient.weight)
  print(patient.married)
  print(patient.allergies)
  print(patient.contact_details)
  print('Updated')


patient_info = {'name':'Ansul','email':'abc@gmail.com','linkedin_url':'http://linkedin.com/312', 'age':23, 'weight':53.5, 'married':False, 'allergies':['pollen','dust'], 'contact_details':{'phone':'9999929999'}}
patient1 = Patient(**patient_info)

insert_patient_data(patient1)
```

```python
from pydantic import BaseModel, EmailStr, AnyUrl, Field
from typing import List, Dict,Optional, Annotated # using annotated and field combination we can attach metadata

class Patient(BaseModel):

  name : Annotated[str, Field(max_length=50, title='Name of the patient',description='give the name of patient less the 50 chars', examples=['Nitish','Ansul'])]
  email : EmailStr
  linkedin_url : AnyUrl
  age : int = Field(gt=18, lt=48) # greater than 18 and less then 48
  weight : Annotated[float, Field(gt=0, strict=True)] # strict type conversion hone se rok deta hai wo smartness supress kr deta hai
  married : Annotated[bool, Field(default=None, description='Is the patient married or not')]
  allergies : Annotated[Optional[List[str]], Field(default=None, max_length=5)]
  contact_details : Dict[str, str]


def insert_patient_data(patient: Patient):

  print(patient.name)
  print(patient.email)
  print(patient.linkedin_url)
  print(patient.age)
  print(patient.weight)
  print(patient.married)
  print(patient.allergies)
  print(patient.contact_details)
  print('Inserted into database')

def update_patient_data(patient: Patient):

  print(patient.name)
  print(patient.email)
  print(patient.age)
  print(patient.weight)
  print(patient.married)
  print(patient.allergies)
  print(patient.contact_details)
  print('Updated')


patient_info = {'name':'Ansul','email':'abc@gmail.com','linkedin_url':'http://linkedin.com/312', 'age':23, 'weight':53.5, 'married':False, 'allergies':['pollen','dust'], 'contact_details':{'phone':'9999929999'}}
patient1 = Patient(**patient_info)

insert_patient_data(patient1)
```
