### Computed Field

* baki ki field use krke ek value calculate knra hai
example BMI

```python
from pydantic import BaseModel, EmailStr, computed_field
from typing import List, Dict

class Patient(BaseModel):

  name : str
  email : EmailStr
  age : int
  weight : float #kg
  height : float #mtr
  married : bool
  allergies : List[str]
  contact_details : Dict[str, str]

  @computed_field
  @property
  def bmi(self) -> float:
    bmi = round(self.weight / (self.height**2),2)
    return bmi

def update_patient_data(patient: Patient):

  print(patient.name)
  print(patient.email)
  print(patient.age)
  print('BMI', patient.bmi)
  print(patient.weight)
  print(patient.married)
  print(patient.allergies)
  print(patient.contact_details)
  print('Updated')


patient_info = {'name':'shon','email':'shon@hdfc.com','linkedin_url':'http://linkedin.com/0804', 'age':'65', 'weight':53.5, 'height':1.73, 'married':False, 'allergies':['pollen','dust'], 'contact_details':{'phone':'9999929999','emergency':'6969696969'}}
patient1 = Patient(**patient_info) # validation -> type coercion

update_patient_data(patient1)
```
